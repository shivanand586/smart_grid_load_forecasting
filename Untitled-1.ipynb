```python
import pandas as pd
```

```python
df=pd.read_csv("/Users/shiv/Downloads/ml_project/LD2011_2014.txt", sep=';', decimal=',')
```

```python
df.head()
```

```python
# Create a copy of the dataset to avoid corrupting the original
df_processed = df.copy()

# Rename the first column for easier handling
timestamp_col = df_processed.columns[0]

# Convert the first column to datetime
df_processed[timestamp_col] = pd.to_datetime(df_processed[timestamp_col])

# Set as index
df_processed.set_index(timestamp_col, inplace=True)

# Create total load
df_processed['load'] = df_processed.sum(axis=1)

# Keep only load column
df_processed = df_processed[['load']]

df_processed.head()
```

```python
df_processed.info()
df_processed.describe()
```

```python
df_processed = df_processed.ffill()
```

```python
df_processed['hour'] = df_processed.index.hour
df_processed['day'] = df_processed.index.dayofweek
df_processed['month'] = df_processed.index.month
df_processed['is_weekend'] = (df_processed['day'] >= 5).astype(int)

df_processed.head()
```

```python
df_processed['lag_1'] = df_processed['load'].shift(1)
df_processed['lag_24'] = df_processed['load'].shift(24)

df_processed = df_processed.dropna()

df_processed.head()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,5))
plt.plot(df_processed['load'])
plt.title("Electric Load Over Time")
plt.show()

df_processed.groupby('hour')['load'].mean().plot()
plt.title("Average Load by Hour")
plt.show()

sns.heatmap(df_processed.corr(), annot=True)
plt.title("Feature Correlation")
plt.show()
```

```python
split = int(0.7 * len(df_processed))

train = df_processed[:split]
test = df_processed[split:]

print(f"Train shape: {train.shape}, Test shape: {test.shape}")
```

```python
X_train = train.drop('load', axis=1)
y_train = train['load']

X_test = test.drop('load', axis=1)
y_test = test['load']

print(f"X_train shape: {X_train.shape}, y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape}, y_test shape: {y_test.shape}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

# Train Linear Regression model
lr = LinearRegression()
lr.fit(X_train, y_train)

# Make predictions
y_pred_lr = lr.predict(X_test)

# Evaluate model
mse = mean_squared_error(y_test, y_pred_lr)
rmse = mse ** 0.5
mae = mean_absolute_error(y_test, y_pred_lr)
r2 = r2_score(y_test, y_pred_lr)

print(f"Linear Regression Results:")
print(f"RMSE: {rmse:.2f}")
print(f"MAE: {mae:.2f}")
print(f"R² Score: {r2:.4f}")
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Train Random Forest model
rf = RandomForestRegressor(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)

# Make predictions
y_pred_rf = rf.predict(X_test)

# Evaluate model
mse_rf = mean_squared_error(y_test, y_pred_rf)
rmse_rf = mse_rf ** 0.5
mae_rf = mean_absolute_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print(f"Random Forest Results:")
print(f"RMSE: {rmse_rf:.2f}")
print(f"MAE: {mae_rf:.2f}")
print(f"R² Score: {r2_rf:.4f}")
```

```python
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Scale features (important for SVM)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train Support Vector Regressor model
svr = SVR(kernel='rbf', C=100, gamma='scale')
svr.fit(X_train_scaled, y_train)

# Make predictions
y_pred_svr = svr.predict(X_test_scaled)

# Evaluate model
mse_svr = mean_squared_error(y_test, y_pred_svr)
rmse_svr = mse_svr ** 0.5
mae_svr = mean_absolute_error(y_test, y_pred_svr)
r2_svr = r2_score(y_test, y_pred_svr)

print(f"Support Vector Regressor Results:")
print(f"RMSE: {rmse_svr:.2f}")
print(f"MAE: {mae_svr:.2f}")
print(f"R² Score: {r2_svr:.4f}")
```

```python
import numpy as np

def evaluate(y_true, y_pred, name):
    print(f"\n{name}")
    print("MAE:", mean_absolute_error(y_true, y_pred))
    print("RMSE:", np.sqrt(mean_squared_error(y_true, y_pred)))
    print("R2:", r2_score(y_true, y_pred))

# Evaluate all models
evaluate(y_test, y_pred_lr, "Linear Regression")
evaluate(y_test, y_pred_rf, "Random Forest")
evaluate(y_test, y_pred_svr, "Support Vector Regressor")
```

```python
evaluate(y_test, y_pred_lr, "Linear Regression")
evaluate(y_test, y_pred_rf, "Random Forest")
evaluate(y_test, y_pred_svr, "SVR")

# Create comparison visualization
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Plot 1: Actual vs Predicted (Linear Regression)
axes[0].scatter(y_test, y_pred_lr, alpha=0.5)
axes[0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
axes[0].set_xlabel('Actual')
axes[0].set_ylabel('Predicted')
axes[0].set_title('Linear Regression')

# Plot 2: Actual vs Predicted (Random Forest)
axes[1].scatter(y_test, y_pred_rf, alpha=0.5, color='green')
axes[1].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
axes[1].set_xlabel('Actual')
axes[1].set_ylabel('Predicted')
axes[1].set_title('Random Forest')

# Plot 3: Actual vs Predicted (SVR)
axes[2].scatter(y_test, y_pred_svr, alpha=0.5, color='orange')
axes[2].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
axes[2].set_xlabel('Actual')
axes[2].set_ylabel('Predicted')
axes[2].set_title('Support Vector Regressor')

plt.tight_layout()
plt.show()

# Create comparison dataframe
comparison_df = pd.DataFrame({
    'Model': ['Linear Regression', 'Random Forest', 'SVR'],
    'MAE': [mean_absolute_error(y_test, y_pred_lr), 
            mean_absolute_error(y_test, y_pred_rf),
            mean_absolute_error(y_test, y_pred_svr)],
    'RMSE': [np.sqrt(mean_squared_error(y_test, y_pred_lr)),
             np.sqrt(mean_squared_error(y_test, y_pred_rf)),
             np.sqrt(mean_squared_error(y_test, y_pred_svr))],
    'R² Score': [r2_score(y_test, y_pred_lr),
                 r2_score(y_test, y_pred_rf),
                 r2_score(y_test, y_pred_svr)]
})

print("\n" + "="*60)
print("MODEL COMPARISON SUMMARY")
print("="*60)
print(comparison_df.to_string(index=False))
```

```python
plt.figure(figsize=(14, 6))
plt.plot(y_test.values, label='Actual', linewidth=2, color='black')
plt.plot(y_pred_lr, label='Linear Regression', linewidth=1.5, alpha=0.7)
plt.plot(y_pred_rf, label='Random Forest', linewidth=1.5, alpha=0.7)
plt.plot(y_pred_svr, label='SVR', linewidth=1.5, alpha=0.7)
plt.legend(loc='best', fontsize=10)
plt.xlabel('Time Index')
plt.ylabel('Electric Load')
plt.title("Actual vs Predicted Load - All Models Comparison")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Calculate and display mean absolute percentage error for better interpretation
def mape(y_true, y_pred):
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100

print(f"\nMean Absolute Percentage Error (MAPE):")
print(f"Linear Regression: {mape(y_test.values, y_pred_lr):.2f}%")
print(f"Random Forest: {mape(y_test.values, y_pred_rf):.2f}%")
print(f"SVR: {mape(y_test.values, y_pred_svr):.2f}%")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_percentage_error as mape

# Assuming y_test, y_pred_lr, y_pred_rf, y_pred_svr are already defined

# Calculate MAPE values dynamically
mape_lr = mape(y_test.values, y_pred_lr)
mape_rf = mape(y_test.values, y_pred_rf)
mape_svr = mape(y_test.values, y_pred_svr)

models = ['Linear Regression', 'Random Forest', 'SVR']
mape_values = [mape_lr, mape_rf, mape_svr]
colors = ['#2ecc71', '#3498db', '#e74c3c']  # Green, Blue, Red

fig, ax = plt.subplots(figsize=(10, 6))
bars = ax.bar(models, mape_values, color=colors, alpha=0.7, edgecolor='black', linewidth=1.5)

# Add value labels on bars
for i, (bar, value) in enumerate(zip(bars, mape_values)):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{value:.2f}%',
            ha='center', va='bottom', fontsize=12, fontweight='bold')

ax.set_ylabel('Mean Absolute Percentage Error (%)', fontsize=12, fontweight='bold')
ax.set_title('Model Comparison - MAPE', fontsize=14, fontweight='bold')
ax.set_ylim(0, max(mape_values) * 1.15)
ax.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()

print("\n" + "="*50)
print("MAPE COMPARISON (Lower is Better)")
print("="*50)
for model, mape_val in zip(models, mape_values):
    print(f"{model:20s}: {mape_val:6.2f}%")
```

```python
future_steps = 24 * 4  # 96 steps
freq = '15min'
```

```python
future_dates = pd.date_range(
    start=df_processed.index[-1],
    periods=future_steps + 1,
    freq=freq
)[1:]
```

```python
future_predictions = []
current_data = df_processed.copy()

for date in future_dates:
    
    new_row = {}

    # Time features
    new_row['hour'] = date.hour
    new_row['day'] = date.dayofweek
    new_row['month'] = date.month
    new_row['is_weekend'] = int(date.dayofweek >= 5)

    # Lag features
    new_row['lag_1'] = current_data['load'].iloc[-1]
    new_row['lag_24'] = current_data['load'].iloc[-24]

    # Predict
    X_new = pd.DataFrame([new_row])
    y_pred = rf.predict(X_new)[0]

    future_predictions.append(y_pred)

    # Append back
    new_row['load'] = y_pred
    new_row_df = pd.DataFrame([new_row], index=[date])
    current_data = pd.concat([current_data, new_row_df])
```

```python
future_df = pd.DataFrame({
    'datetime': future_dates,
    'predicted_load': future_predictions
}).set_index('datetime')
```

```python
plt.figure(figsize=(14,6))

# Plot last 3 days of real data
history_window = 72  # last 72 hours

plt.plot(df_processed['load'][-history_window:], 
         label='Historical Load', 
         color='blue')

# Plot forecast (bold + different style)
plt.plot(future_df['predicted_load'], 
         label='24-Hour Forecast', 
         color='red', 
         linestyle='dashed', 
         linewidth=2.5)

# Highlight boundary line (VERY IMPORTANT)
plt.axvline(x=df_processed.index[-1], 
            color='black', 
            linestyle='--', 
            label='Forecast Start')

plt.title("Electric Load Forecast (Next 24 Hours)", fontsize=14)
plt.xlabel("Time")
plt.ylabel("Load")
plt.legend()
plt.grid(alpha=0.3)

plt.show()
```

```python
# Complete forecast visualization with confidence intervals
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

fig, ax = plt.subplots(figsize=(16, 7))

# Plot last 3 days of real data
history_window = 72  # last 72 hours
historical_data = df_processed['load'][-history_window:]
ax.plot(historical_data.index, historical_data.values, 
        label='Historical Load', color='blue', linewidth=2)

# Plot forecast
ax.plot(future_df.index, future_df['predicted_load'].values, 
        label='24-Hour Forecast', color='red', linestyle='dashed', linewidth=2.5)

# Add confidence interval (2% uncertainty)
uncertainty = 0.02
upper = future_df['predicted_load'] * (1 + uncertainty)
lower = future_df['predicted_load'] * (1 - uncertainty)
ax.fill_between(future_df.index, lower.values, upper.values, 
                color='red', alpha=0.2, label='Confidence Interval (±2%)')

# Highlight boundary line
ax.axvline(x=df_processed.index[-1], 
           color='black', linestyle='--', linewidth=2,
           label='Forecast Start')

# Format x-axis
ax.xaxis.set_major_locator(mdates.HourLocator(interval=6))
ax.xaxis.set_major_formatter(mdates.DateFormatter('%m-%d %H:%M'))
plt.xticks(rotation=45, ha='right')

# Add text annotation
ax.text(0.5, 0.95, '24-Hour Forecast Horizon', 
        transform=ax.transAxes, fontsize=12, color='red', fontweight='bold',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.8),
        ha='center', va='top')

ax.set_title("Electric Load Forecast - Next 24 Hours with Confidence Interval", 
             fontsize=14, fontweight='bold')
ax.set_xlabel("Time", fontsize=12)
ax.set_ylabel("Electric Load (kW)", fontsize=12)
ax.legend(loc='best', fontsize=11)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Print forecast summary
print("\n" + "="*70)
print("FORECAST SUMMARY")
print("="*70)
print(f"Forecast Period: {future_df.index[0]} to {future_df.index[-1]}")
print(f"Number of Forecasted Points: {len(future_df)}")
print(f"Average Forecasted Load: {future_df['predicted_load'].mean():.2f} kW")
print(f"Max Forecasted Load: {future_df['predicted_load'].max():.2f} kW")
print(f"Min Forecasted Load: {future_df['predicted_load'].min():.2f} kW")
print(f"Standard Deviation: {future_df['predicted_load'].std():.2f} kW")
print(f"Confidence Interval (±2%): {future_df['predicted_load'].min() * 0.98:.2f} - {future_df['predicted_load'].max() * 1.02:.2f} kW")
print("="*70)
```
